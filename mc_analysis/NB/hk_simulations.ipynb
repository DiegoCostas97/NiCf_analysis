```python
import numpy as np
import matplotlib.pyplot as plt

# ========================
# DATOS
# ========================
hits_000    = df_digiHits_0.groupby("event_id").size().values
hits_2000   = df_digiHits_2000.groupby("event_id").size().values
hits_m2000  = df_digiHits_minus2000.groupby("event_id").size().values
hits_z2000  = df_digiHits_z2000.groupby("event_id").size().values
hits_zm2000 = df_digiHits_zminus2000.groupby("event_id").size().values

all_hits = [hits_000, hits_2000, hits_m2000, hits_z2000, hits_zm2000]

# ========================
# MEDIA Y STD SOLO PARA 000
# ========================
mean_000 = np.mean(hits_000)
std_000  = np.std(hits_000)

label_000 = rf"(0, 0, 0)" + "\n" + f"mean {mean_000:.1f}" +"\n" + f"std {std_000:.1f}"

# ========================
# BINNING COMÚN
# ========================
nbins = 60
xmin  = 0
xmax  = max(np.max(h) for h in all_hits)
bins  = np.linspace(xmin, xmax, nbins + 1)
centers = 0.5 * (bins[1:] + bins[:-1])

# --- Ratio wrt (0,0,0) ---
def safe_ratio(h, href):
    return np.divide(h, href, out=np.full_like(h, np.nan), where=href > 0)

# ========================
# HISTOGRAMAS NORMALIZADOS
# ========================
def hist(data):
    return np.histogram(data, bins=bins, density=True)[0]

h_000    = hist(hits_000)
h_2000   = hist(hits_2000)
h_m2000  = hist(hits_m2000)
h_z2000  = hist(hits_z2000)
h_zm2000 = hist(hits_zm2000)

def ratio(h, href):
    return np.divide(h, href, out=np.full_like(h, np.nan), where=href > 0)

# ========================
# PLOT
# ========================
fig, (ax, rax) = plt.subplots(
    2, 1, figsize=(8, 6),
    sharex=True, gridspec_kw={"height_ratios": [3, 1]}
)

# --- Histos ---
ax.step(centers, h_000,    where="mid", lw=2, label=label_000)
ax.step(centers, h_2000,   where="mid", label="(0, 2000, 0)")
ax.step(centers, h_m2000,  where="mid", label="(0, -2000, 0)")
ax.step(centers, h_z2000,  where="mid", label="(0, 0, 2000)")
ax.step(centers, h_zm2000, where="mid", c="purple", label="(0, 0, -1000)")

ax.set_ylabel("Normalized entries")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)

# --- Ratio ---
rax.step(centers, safe_ratio(h_2000,   h_000), where="mid")
rax.step(centers, safe_ratio(h_m2000,  h_000), where="mid")
rax.step(centers, safe_ratio(h_z2000,  h_000), where="mid")
rax.step(centers, safe_ratio(h_zm2000, h_000), where="mid")

rax.axhline(1.0, color="k", ls="--", lw=1)
# rax.set_ylim(0.8, 1.2)
rax.set_ylabel("Ratio to (0,0,0)")
rax.set_xlabel("No. Hits Per Event")
rax.grid(alpha=0.3)
# rax.legend();

plt.tight_layout()
plt.show()
```

```python
# ========================
# DATOS (radios)
# ========================
hits_9cm  = df_digiHits_9cm_000.groupby("event_id").size().values
hits_12cm = df_digiHits_12cm_000.groupby("event_id").size().values
hits_15cm = df_digiHits_15cm_000.groupby("event_id").size().values

all_hits_r = [hits_9cm, hits_12cm, hits_15cm]

mean_9cm = np.mean(hits_9cm)
std_9cm  = np.std(hits_9cm)

label_9cm = rf"(0, 0, 0) - 9 cm" + "\n" + f"mean {mean_9cm:.1f}" +"\n" + f"std {std_9cm:.1f}"

mean_12cm = np.mean(hits_12cm)
std_12cm  = np.std(hits_12cm)

label_12cm = rf"(0, 0, 0) - 12 cm" + "\n" + f"mean {mean_12cm:.1f}" +"\n" + f"std {std_12cm:.1f}"

mean_15cm = np.mean(hits_15cm)
std_15cm  = np.std(hits_15cm)

label_15cm = rf"(0, 0, 0) - 15 cm" + "\n" + f"mean {mean_15cm:.1f}" +"\n" + f"std {std_15cm:.1f}"

# ========================
# BINNING COMÚN
# ========================
nbins = 60
xmin  = 0
xmax  = max(np.max(h) for h in all_hits_r)
bins  = np.linspace(xmin, xmax, nbins + 1)
centers = 0.5 * (bins[1:] + bins[:-1])

# ========================
# HISTOGRAMAS
# ========================
h_9cm  = np.histogram(hits_9cm,  bins=bins, density=True)[0]
h_12cm = np.histogram(hits_12cm, bins=bins, density=True)[0]
h_15cm = np.histogram(hits_15cm, bins=bins, density=True)[0]

# ========================
# PLOT
# ========================
fig, (ax, rax) = plt.subplots(
    2, 1, figsize=(8, 6),
    sharex=True, gridspec_kw={"height_ratios": [3, 1]}
)

# --- Histos ---
ax.step(centers, h_9cm,  where="mid", c="red", lw=2, label=label_9cm)
ax.step(centers, h_12cm, where="mid", c="blue", label=label_12cm)
ax.step(centers, h_15cm, where="mid", c="green", label=label_15cm)

ax.set_ylabel("Normalized entries")
ax.legend()
ax.grid(alpha=0.3)

# --- Ratio wrt 9 cm ---
rax.step(centers, safe_ratio(h_9cm, h_000), where="mid", c="red", label="9 cm / 6.75 cm")
rax.step(centers, safe_ratio(h_12cm, h_000), where="mid", c="blue", label="12 cm / 6.75 cm")
rax.step(centers, safe_ratio(h_15cm, h_000), where="mid", c="green", label="15 cm / 6.75 cm")

rax.axhline(1.0, color="k", ls="--", lw=1)
# rax.set_ylim(0.8, 1.2)
rax.set_ylabel("Ratio to 6.75 cm")
rax.set_xlabel("No. Hits Per Event")
rax.grid(alpha=0.3)
rax.legend();

plt.tight_layout()
plt.show()
```
