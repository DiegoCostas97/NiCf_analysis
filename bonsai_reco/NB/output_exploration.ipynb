```python
# data = z_1767_chargeCut
name= 'gaus'
nbins, size = 100, 120
data = df_1766_charge

subplot = pltext.canvas(3)

subplot(1)
z_results = pltext.hfit(data["z"], nbins, name, xylabels="Z", stats = True, residuals = False, range=(40, 120));

subplot(2)
x_results = pltext.hfit(data["x"], nbins, name, xylabels="X", stats = True, residuals = False, range=(-60, 50));

subplot(3)
y_results = pltext.hfit(data["y"], nbins, name, xylabels="Y", stats = True, residuals = False, range=(-50, 60));
```

# Easy Cluster Selection

```python
# Centro de la esfera
x0, y0, z0 = x_results[3][1], y_results[3][1], z_results[3][1] # reemplaza con tus valores
R = 15  # radio de la esfera

# Calcular distancia al centro
dist2 = (df_1766_charge["x"].to_numpy() - x0)**2 + (df_1766_charge["y"].to_numpy() - y0)**2 + (df_1766_charge["z"].to_numpy() - z0)**2

# Filtrar puntos dentro de la esfera
mask = dist2 <= R**2

x_in = df_1766_charge["x"].to_numpy()[mask]
y_in = df_1766_charge["y"].to_numpy()[mask]
z_in = df_1766_charge["z"].to_numpy()[mask]
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

ax.scatter(x_in, y_in, z_in, alpha=0.1)

ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_zlabel('Z Label');
```
